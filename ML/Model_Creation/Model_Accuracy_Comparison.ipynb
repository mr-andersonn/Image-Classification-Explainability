# Model Accuracy Comparison
Comparison of test accuracies across different models

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Model accuracies
models = ['Random Chance', 'ConvNet_Simple', 'ConvNet_Final', 'MobileNetV3Large_Simple', 'MobileNetV3Large_Final']
accuracies = [0.1111, 0.8389, 0.9189, 0.2589, 0.9972]

# Create bar chart
plt.figure(figsize=(12, 6))
bars = plt.bar(models, accuracies, color=['gray', 'skyblue', 'dodgerblue', 'lightcoral', 'green'])

# Add value labels on top of bars
for i, (bar, acc) in enumerate(zip(bars, accuracies)):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, 
             f'{acc:.4f}', ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.xlabel('Model', fontsize=12, fontweight='bold')
plt.ylabel('Test Accuracy', fontsize=12, fontweight='bold')
plt.title('Model Performance Comparison', fontsize=14, fontweight='bold')
plt.ylim(0, 1.05)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()
```
